```python
print("Enter marks obtained in the five subjects (out of 30):")

user_marks = []

for subject in features:
    while True:
        mark = float(input(f"Enter marks for {subject} (0-30): "))

        if 0 <= mark <= 30:
            user_marks.append(mark)
            break
        else:
            print("Please enter marks between 0 and 30.")
```

```python
user_data = pd.DataFrame(
    [user_marks],
    columns=features
)

print("\nUser Entered Marks:")
display(user_data)
```

```python
user_normalized = scaler.transform(user_data)

user_normalized = pd.DataFrame(
    user_normalized,
    columns=features
)

print("Normalized User Marks:")
display(user_normalized)
```

```python
prediction_euclidean = knn_euclidean.predict(user_normalized)

print(
    "Predicted Grade using Euclidean Distance:",
    prediction_euclidean[0]
)
```

```python
prediction_manhattan = knn_manhattan.predict(user_normalized)

print(
    "Predicted Grade using Manhattan Distance:",
    prediction_manhattan[0]
)
```

```python
prediction_minkowski = knn_minkowski.predict(user_normalized)

print(
    "Predicted Grade using Minkowski Distance:",
    prediction_minkowski[0]
)
```
