### Introduction
K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for both classification and regression problems. It is a simple, instance-based learning algorithm that makes predictions based on the similarity between data points.

In KNN classification, the class of a new data point is determined by examining its K nearest data points in the training dataset. The class that occurs most frequently among the nearest neighbors is assigned to the new data point.

In this practical, KNN is applied to the processed student dataset. The normalized marks of five subjects are used as input features, while the student's grade is used as the target variable.

###K-Nearest Neighbors (KNN)
K-Nearest Neighbors (KNN) is a supervised, non-parametric machine learning algorithm used for classification and regression.

KNN does not build an explicit mathematical model. Instead, it stores the available labeled data and makes a prediction for a new data point by finding its nearest neighbors based on a selected distance measure.

- ### The basic steps of KNN are:

1.Select the value of K.

2.Use the complete labeled dataset as the training data.

3.Calculate the distance between the new input and all training samples.

4.Identify the K nearest neighbors.

5.Determine the most frequent class among the nearest neighbors.

6.Assign that class to the new input.

In this practical, K = 5 is used with three distance measures: Euclidean, Manhattan, and Minkowski distance with (p=3).

The complete processed student dataset is used for training, and a new student's subject marks are entered by the user for grade prediction.

```python
X = df[['SOFTWARE ENGINEER',
        'TOC',
        'COMPUTER NETWORK',
        'MACHINE LEARNING',
        'DATA WAREHOUSE']]
```

```python
y = df['GRADE']
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

### 1. Euclidean Distance
Euclidean distance measures the straight-line distance between two points. It is commonly used when the distance between data points is measured in a geometric space.

```python
knn_euclidean = KNeighborsClassifier(
    n_neighbors=5,
    metric='euclidean'
)

knn_euclidean.fit(X, y)
```

###2. Manhattan Distance
Manhattan distance calculates the distance between two points by adding the absolute differences between their corresponding features.

```python
knn_manhattan = KNeighborsClassifier(
    n_neighbors=5,
    metric='manhattan'
)

knn_manhattan.fit(X, y)
```

###3. Minkowski Distance
Minkowski distance is a generalized distance measure that can represent different distance metrics depending on the value of (p).

```python
knn_minkowski = KNeighborsClassifier(
    n_neighbors=5,
    metric='minkowski',
    p=3
)

knn_minkowski.fit(X, y)
```

###User Input and Grade Prediction
The complete processed dataset is used to train the KNN models.

The user is asked to enter marks obtained in the five subjects. Since the trained dataset contains normalized marks, the entered marks are converted to the normalized scale before prediction.

The trained KNN models then predict the grade of the new student using Euclidean, Manhattan, and Minkowski distance measures.

```python
print("Enter marks obtained out of 60")
print()

software_engineer = float(input("Software Engineering: "))
toc = float(input("Theory of Computation (TOC): "))
computer_network = float(input("Computer Network: "))
machine_learning = float(input("Machine Learning: "))
data_warehouse = float(input("Data Warehouse: "))
```

```python
marks = [
    software_engineer,
    toc,
    computer_network,
    machine_learning,
    data_warehouse
]

if any(mark < 0 or mark > 60 for mark in marks):
    raise ValueError("Marks must be between 0 and 60.")
```

```python
new_student = np.array([[
    software_engineer / 60,
    toc / 60,
    computer_network / 60,
    machine_learning / 60,
    data_warehouse / 60
]])

new_student
```

###Prediction
Here for prediction, a new student's subject marks are entered by the user. The trained KNN models calculate the nearest neighbors of this new data point and predict its grade using Euclidean, Manhattan, and Minkowski distance measures.

The final predicted grade is obtained using majority voting among the three distance-based predictions.

```python
prediction_euclidean = knn_euclidean.predict(new_student)[0]

prediction_manhattan = knn_manhattan.predict(new_student)[0]

prediction_minkowski = knn_minkowski.predict(new_student)[0]
```

```python
print("Predicted Grade")
print("-------------------------")
print("Euclidean Distance :", prediction_euclidean)
print("Manhattan Distance :", prediction_manhattan)
print("Minkowski Distance :", prediction_minkowski)
```

```python
from collections import Counter

predictions = [
    prediction_euclidean,
    prediction_manhattan,
    prediction_minkowski
]

final_prediction = Counter(predictions).most_common(1)[0][0]

print("\nFinal Predicted Grade:", final_prediction)
```
