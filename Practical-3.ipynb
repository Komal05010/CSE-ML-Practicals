```python
K = 5

euclidean_neighbours = np.argsort(
    euclidean_matrix,
    axis=1
)[:, :K]
```

```python
print("Nearest neighbours using Euclidean distance:")
print(euclidean_neighbours)
```

```python
import numpy as np

euclidean_labels = np.asarray(y_train)[euclidean_neighbours]

print("Classes of nearest neighbours:")
print(euclidean_labels)
```

```python
euclidean_predictions = []

for neighbours in euclidean_labels:
    prediction = np.bincount(neighbours).argmax()
    euclidean_predictions.append(prediction)

print("Euclidean predictions:")
print(euclidean_predictions)
```

```python
from sklearn.metrics import accuracy_score

euclidean_accuracy = accuracy_score(
    y_test,
    euclidean_predictions
)

print("Euclidean KNN Accuracy:", euclidean_accuracy)
print("Euclidean KNN Accuracy (%):", euclidean_accuracy * 100)
```

```python
K = 5

manhattan_neighbours = np.argsort(
    manhattan_matrix,
    axis=1
)[:, :K]

print("Manhattan nearest neighbours:")
print(manhattan_neighbours)
```

```python
manhattan_labels = np.asarray(y_train)[manhattan_neighbours]

print("Classes of Manhattan nearest neighbours:")
print(manhattan_labels)
```

```python
manhattan_predictions = []

for neighbours in manhattan_labels:
    prediction = np.bincount(neighbours).argmax()
    manhattan_predictions.append(prediction)

print("Manhattan predictions:")
print(manhattan_predictions)
```

```python
manhattan_accuracy = accuracy_score(
    y_test,
    manhattan_predictions
)

print("Manhattan KNN Accuracy:", manhattan_accuracy)
print("Manhattan KNN Accuracy (%):", manhattan_accuracy * 100)
```

```python
minkowski_neighbours = np.argsort(
    minkowski_matrix,
    axis=1
)[:, :K]

print("Minkowski nearest neighbours:")
print(minkowski_neighbours)
```

```python
minkowski_labels = np.asarray(y_train)[minkowski_neighbours]

print("Classes of Minkowski nearest neighbours:")
print(minkowski_labels)
```

```python
minkowski_predictions = []

for neighbours in minkowski_labels:
    prediction = np.bincount(neighbours).argmax()
    minkowski_predictions.append(prediction)

print("Minkowski predictions:")
print(minkowski_predictions)
```
