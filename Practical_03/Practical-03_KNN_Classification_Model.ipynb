# Practical No. 3: Implementation of K-Nearest Neighbors (KNN)

## Aim
To implement the K-Nearest Neighbors (KNN) classification algorithm on the processed dataset using different distance measures and evaluate its performance.

```python
import pandas as pd
import numpy as np

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## Dataset Used

The processed dataset obtained from Practical 2 is used as the input dataset.

The five normalized subject marks used as input features (`X`) are:

- Software Engineer
- Theory of Computation (TOC)
- Computer Network
- Machine Learning
- Data Warehouse

The `GRADE` column is used as the target variable (`y`).

The dataset contains **61 student records** and **5 input features**.

```python
# Read Excel file
filename = list(uploaded.keys())[0]
df = pd.read_excel(filename)
```

```python
print("Dataset shape:", df.shape)
```

### K-Nearest Neighbors (KNN)

K-Nearest Neighbors is a supervised, non-parametric machine learning algorithm. It does not build an explicit mathematical model during training. Instead, it stores the training data and makes predictions by finding the nearest data points to a given test point.

The basic steps of KNN are:

1. Choose the value of K.
2. Calculate the distance between the test point and all training points.
3. Identify the K nearest neighbors.
4. Determine the most common class among these neighbors.
5. Assign that class to the test point.

In this practical, **K = 7** is used for all three distance measures to ensure a fair comparison.

```python
# Input features
X = df[
    [
        "SOFTWARE ENGINEER",
        "TOC",
        "COMPUTER NETWORK",
        "MACHINE LEARNING",
        "DATA WAREHOUSE"
    ]
]

# Target
y = df["GRADE"]
```

```python
print("\nFeatures (X):")
display(X.head())

print("\nTarget (y):")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
k = 7

knn = KNeighborsClassifier(n_neighbors=k)

knn.fit(X_train, y_train)

print("KNN model trained successfully!")
print("K =", k)
```

## 1. Euclidean Distance

Euclidean distance measures the straight-line distance between two points. It is commonly used when the distance between data points is measured in a geometric space.

**Formula:**

$$
d(x,y) =
\sqrt{
\sum_{i=1}^{n}(x_i-y_i)^2
}
$$

In Scikit-learn:

`metric='euclidean'`


```python
knn_euclidean = KNeighborsClassifier(
    n_neighbors=5,
    metric='euclidean'
)
knn_euclidean.fit(X_train, y_train)
y_pred_euclidean = knn_euclidean.predict(X_test)
```

```python
accuracy_euclidean = accuracy_score(y_test, y_pred_euclidean)
print("Euclidean Distance Accuracy:",
      accuracy_euclidean * 100, "%")
```

### Confusion Matrix

A confusion matrix is a table used to evaluate the classification performance of a model. It shows the number of correctly and incorrectly classified samples for each class.

The rows represent the actual classes, while the columns represent the predicted classes.

### Classification Report

The classification report provides:

- **Precision** – proportion of predicted samples that are actually correct.
- **Recall** – proportion of actual samples correctly identified.
- **F1-score** – harmonic mean of precision and recall.
- **Support** – number of actual samples belonging to each class.

```python
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_euclidean))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_euclidean, zero_division=0))
```

## 2. Manhattan Distance

Manhattan distance calculates the distance between two points by adding the absolute differences between their corresponding features.

**Formula:**

$$
d(x,y) =
\sum_{i=1}^{n}|x_i-y_i|
$$

In Scikit-learn:

`metric='manhattan'`


```python
knn_manhattan = KNeighborsClassifier(
    n_neighbors=5,
    metric='manhattan'
)
knn_manhattan.fit(X_train, y_train)
y_pred_manhattan = knn_manhattan.predict(X_test)
```

```python
accuracy_manhattan = accuracy_score(y_test, y_pred_manhattan)
print("Manhattan Distance Accuracy:",
      accuracy_manhattan * 100, "%")
```

```python
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_manhattan))
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_manhattan,
    zero_division=0
))
```

## 3. Minkowski Distance

Minkowski distance is a generalized distance measure that can represent different distance metrics depending on the value of \(p\).

**Formula:**

$$
d(x,y) =
\left(
\sum_{i=1}^{n}|x_i-y_i|^p
\right)^{1/p}
$$

The value of \(p\) determines the type of distance:

* \(p = 3\) → Minkowski distance used in this practical

In Scikit-learn:

`metric='minkowski', p=3`


```python
knn_minkowski = KNeighborsClassifier(
    n_neighbors=5,
    metric='minkowski',
    p=3
)
knn_minkowski.fit(X_train, y_train)
y_pred_minkowski = knn_minkowski.predict(X_test)
```

```python
accuracy_minkowski = accuracy_score(y_test, y_pred_minkowski)
print("Minkowski Distance Accuracy:",
      accuracy_minkowski * 100, "%")
```

```python
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_minkowski))
print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_minkowski,
    zero_division=0
))
```

### Accuracy

Accuracy represents the proportion of correctly classified samples out of all test samples.

$$
Accuracy =
\frac{Correct\ Predictions}
{Total\ Predictions}
$$


```python
results = pd.DataFrame({
    'Distance': ['Euclidean', 'Manhattan', 'Minkowski'],
    'Accuracy': [
        accuracy_euclidean,
        accuracy_manhattan,
        accuracy_minkowski
    ]
})
results['Accuracy (%)'] = results['Accuracy'] * 100
results
```

```python
best_distance = results.loc[
    results['Accuracy'].idxmax(),
    'Distance'
]
best_accuracy = results['Accuracy (%)'].max()
print("Best Distance:", best_distance)
print("Best Accuracy:", best_accuracy, "%")
```

### Conclusion

K-Nearest Neighbors (KNN) was successfully implemented on the processed student dataset using three different distance measures: Euclidean, Manhattan, and Minkowski distance. The five normalized subject marks were used as input features and `GRADE` was used as the target variable.
