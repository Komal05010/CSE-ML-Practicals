# Practical 01: Pandas Import and Export

## Aim
To perform data import and export operations using the Pandas library in Python with different file formats such as Excel, CSV, JSON, and TSV.

## 1. Importing the Required Library

```python
import pandas as pd
```

```python
from google.colab import files
uploaded = files.upload()
```

## 2. Importing Data from Excel

The `read_excel()` function is used to read data from an Excel file and store it in a Pandas DataFrame.

```python
df = pd.read_excel(
    "Mldata.xlsx",
    dtype={"ROLL NO.":"string"}
    )
```

```python
print(df)
```

## 3. Exporting Data to CSV

The `to_csv()` function is used to export the DataFrame into a CSV (Comma-Separated Values) file.



```python
df.to_csv("exported_data.csv",
index = False)
```

## 4. Exporting Data to Excel

The `to_excel()` function is used to export a Pandas DataFrame into an Excel file.

```python
df.to_excel("exported_data.xlsx",
index = False)
```
