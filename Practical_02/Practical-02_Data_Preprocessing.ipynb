## HANDLE MISSING VALUES

Missing values in numerical columns are handled using the **median / mode imputation** technique.

The median / mode of the valid values in each numerical subject column is calculated and used to replace the missing values.

This helps retain the available records without unnecessarily removing rows from the dataset.

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
for column in categorical_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print(df.isnull().sum())
```

## CLEAN INCONSISTENT CATEGORICAL DATA

Categorical columns are examined for inconsistent or unwanted values.

The **GENDER** column is cleaned and standardized so that only valid categorical values are retained.

After cleaning, the unique gender values are checked to ensure consistency.

```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
print(df["GENDER"].unique())
```

```python
df["GENDER"] = df["GENDER"].astype("string").str.strip().str.lower()

df["GENDER"] = df["GENDER"].replace({
    "m": "Male",
    "male": "Male",
    "mens":"Male",
    "f": "Female",
    "female": "Female"
})

print(df["GENDER"].unique())
```

```python
print(df["GENDER"].value_counts(dropna=False))
```

## PROCESS GRADE AND RESULT

The **GRADE** and **RESULT** columns are retained as part of the dataset.

Where their values need to be generated or corrected, they are assigned according to the specified grading criteria and result conditions based on the students' marks.

This ensures that the final dataset contains consistent and meaningful grade and result information.

```python
print(df["GRADE"].unique())
```

```python
# Clean the GRADE column
df["GRADE"] = df["GRADE"].astype("string").str.strip().str.upper()

# Define valid grades
valid_grades = ["A", "B", "A+", "C", "P", "B+", "F"]

# Keep valid grades and replace everything else with F
df["GRADE"] = df["GRADE"].where(df["GRADE"].isin(valid_grades), "F")

# Display the cleaned grades
print(df["GRADE"].unique())
```

```python
print(df["RESULT"].unique())
```

```python
df["RESULT"] = df["RESULT"].astype("string").str.strip().str.upper()
df["RESULT"] = df["RESULT"].where(df["RESULT"] == "PASS", "FAIL")

print(df["RESULT"].unique())
```

## CHECK AND REMOVE DUPLICATE RECORDS

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Duplicate rows after removal:", df.duplicated().sum())
```

## DETECT AND HANDLE OUTLIERS

Outliers are observations that differ significantly from the general pattern of the data.

The numerical subject columns are examined for unusually high or low values. An observed maximum value is checked against the valid marks range before deciding whether it represents a genuine observation or an error.

Valid observations are not removed simply because they are relatively high or low.

```python
df_encoded = pd.get_dummies(
    df,
    columns=["GENDER"],
    drop_first=True
)

display(df_encoded.head())
```

```python
numeric_columns = df_encoded.select_dtypes(
    include=np.number
).columns

print(numeric_columns)
```

```python
for column in numeric_columns:

    Q1 = df_encoded[column].quantile(0.25)
    Q3 = df_encoded[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_encoded[
        (df_encoded[column] < lower_bound) |
        (df_encoded[column] > upper_bound)
    ]

    print(column, ":", len(outliers), "potential outliers")
```

```python
df_clean = df_encoded.copy()

for column in numeric_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[column] >= lower_bound) &
        (df_clean[column] <= upper_bound)
    ]
```

```python
print("Rows before outlier handling:", len(df_encoded))
print("Rows after outlier handling:", len(df_clean))
```

## FEATURE SCALING

**A. Standardization**

Standardization changes numerical features to approximately:

```python
standardized_data = df_clean.copy()

numeric_columns = standardized_data.select_dtypes(
    include=np.number
).columns

scaler = StandardScaler()

standardized_data[numeric_columns] = scaler.fit_transform(
    standardized_data[numeric_columns]
)

display(standardized_data.head())
```

**B.  MIN-MAX Normalization**

Now demonstrate Min-Max normalization.

```python
normalized_data = df_clean.copy()

numeric_columns = normalized_data.select_dtypes(
    include=np.number
).columns
minmax_scaler = MinMaxScaler()

normalized_data[numeric_columns] = minmax_scaler.fit_transform(
    normalized_data[numeric_columns]
)

display(normalized_data.head())
```
