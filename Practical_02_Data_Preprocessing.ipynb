```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel(
    "DATASETS/Mldata.xlsx",
    dtype={"ROLL NO.": "string"}
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

## EXPLORE THE DATASET

The dataset is initially explored to understand its structure and contents.

The following information is examined:

- Number of rows and columns
- Column names
- Data types
- First few records
- Basic statistical information

This initial exploration helps identify possible data quality issues that need to be addressed during preprocessing.

```python
print("First 5 rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

## CHECK MISSING VALUES

Missing values are checked in each column of the dataset.

Identifying missing values is necessary because they may affect statistical calculations and the performance of machine learning algorithms.

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("\nMissing value percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
# Clean column names by removing leading/trailing spaces

df.columns = df.columns.str.strip()

print("Cleaned column names:")
print(df.columns.tolist())
```

```python
subject_columns = [
    "SOFTWARE ENGINEER",
    "TOC",
    "COMPUTER NETWORK",
    "MACHINE LEARNING",
    "DATA WAREHOUSE"
]

for col in subject_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Subject columns converted to numeric.")
```

## HANDLE INVALID OR EMPTY VALUES

The dataset is checked for empty, invalid, or inconsistent entries.

Where required, such values are cleaned or replaced with appropriate values so that the dataset remains consistent and suitable for further processing.

```python
before = len(df)

df = df.dropna(how="all")

after = len(df)

print("Completely empty rows removed:", before - after)
print("New shape:", df.shape)
```

```python
before = len(df)

df = df.dropna(subset=["ROLL NO."])

df = df.dropna(
    subset=subject_columns,
    how="all"
)

after = len(df)

print("Invalid/empty records removed:", before - after)
print("New shape:", df.shape)
```

## HANDLE MISSING VALUES

Missing values in numerical columns are handled using the **mean imputation** technique.

The mean of the valid values in each numerical subject column is calculated and used to replace the missing values.

This helps retain the available records without unnecessarily removing rows from the dataset.

```python
for col in subject_columns:
    mean_value = df[col].mean()
    df[col] = df[col].fillna(mean_value)

print("Missing numerical values filled with subject-wise mean.")
```

```python
print("Missing values after mean imputation:")
print(df.isnull().sum())
```

```python
print("Current dataset shape:", df.shape)

display(df.head(15))
```

```python
print("\nDataset information after preprocessing:")
df.info()
```

## DETECT AND HANDLE OUTLIERS

Outliers are observations that differ significantly from the general pattern of the data.

The numerical subject columns are examined for unusually high or low values. An observed maximum value is checked against the valid marks range before deciding whether it represents a genuine observation or an error.

Valid observations are not removed simply because they are relatively high or low.

```python
# Check for invalid marks outside the valid range 0–60

for col in subject_columns:
    invalid = df[(df[col] < 0) | (df[col] > 60)][["ROLL NO.", col]]

    if not invalid.empty:
        print(f"\nInvalid values in {col}:")
        display(invalid)
```

```python
# Calculate mean using only valid marks (0 to 60)

valid_means = {}

for col in subject_columns:
    valid_values = df.loc[
        (df[col] >= 0) & (df[col] <= 60),
        col
    ]

    valid_means[col] = valid_values.mean()

print("Mean of valid marks for each subject:\n")

for col, mean in valid_means.items():
    print(f"{col}: {mean:.2f}")
```

```python
# Replace invalid marks (<0 or >60) with the mean of valid marks

for col in subject_columns:
    mean_value = valid_means[col]

    invalid_mask = (df[col] < 0) | (df[col] > 60)

    df.loc[invalid_mask, col] = mean_value

print("Invalid marks replaced with subject-wise valid mean.")
```

```python
print("Checking for invalid marks after replacement:\n")

for col in subject_columns:
    invalid_count = ((df[col] < 0) | (df[col] > 60)).sum()

    print(f"{col}: {invalid_count} invalid values")
```

```python
# Display records containing the previously invalid marks

invalid_roll_numbers = [
    "240104010056",
    "240104010032",
    "240104010061",
    "240104010024",
    "240104010046"
]

display(
    df[df["ROLL NO."].isin(invalid_roll_numbers)]
)
```

## CLEAN CATEGORICAL DATA

Categorical columns are examined for inconsistent or unwanted values.

The **GENDER** column is cleaned and standardized so that only valid categorical values are retained.

After cleaning, the unique gender values are checked to ensure consistency.

```python
# Clean and standardize gender values

df["GENDER"] = df["GENDER"].str.strip().str.upper()

gender_mapping = {
    "M": "M",
    "MALE": "M",
    "MEN": "M",
    "MENS": "M",
    "F": "F",
    "FEMALE": "F"
}

df["GENDER"] = df["GENDER"].map(gender_mapping)

print("Gender values after cleaning:")
print(df["GENDER"].value_counts())
```

```python
print("Unique Gender values:")
print(df["GENDER"].unique())
```

## PROCESS GRADE AND RESULT

The **GRADE** and **RESULT** columns are retained as part of the dataset.

Where their values need to be generated or corrected, they are assigned according to the specified grading criteria and result conditions based on the students' marks.

This ensures that the final dataset contains consistent and meaningful grade and result information.

```python
# Calculate average marks across all five subjects

df["average_marks"] = df[subject_columns].mean(axis=1)

display(
    df[["ROLL NO."] + subject_columns + ["average_marks"]].head(10)
)
```

```python
# Recalculate Grade based on average marks

def calculate_grade(average):
    if average >= 54:
        return "O"
    elif average >= 48:
        return "A+"
    elif average >= 42:
        return "A"
    elif average >= 36:
        return "B"
    elif average >= 30:
        return "C"
    elif average >= 24:
        return "P"
    else:
        return "F"

df["GRADE"] = df["average_marks"].apply(calculate_grade)

print("Grade recalculated successfully.")
```

```python
# Recalculate Result based on average marks

df["RESULT"] = np.where(
    df["average_marks"] >= 24,
    "PASS",
    "FAIL"
)

print("Result recalculated successfully.")
```

```python
df.drop(columns=["average_marks"], inplace=True)

print("Temporary average column removed.")
print("Current shape:", df.shape)
```

```python
print("Grade distribution:")
print(df["GRADE"].value_counts().sort_index())

print("\nResult distribution:")
print(df["RESULT"].value_counts())

display(df)
```

```python
# Check for completely duplicate records

duplicate_count = df.duplicated().sum()

print("Number of completely duplicate records:", duplicate_count)
```

```python
# Display completely duplicate records

duplicates = df[df.duplicated(keep=False)]

if duplicates.empty:
    print("No completely duplicate records found.")
else:
    display(duplicates)
```

```python
# Encode target variable
df['RESULT'] = df['RESULT'].map({'PASS': 1, 'FAIL': 0})
```

```python
from sklearn.preprocessing import MinMaxScaler

features = [
    'SOFTWARE ENGINEER',
    'TOC',
    'COMPUTER NETWORK',
    'MACHINE LEARNING',
    'DATA WAREHOUSE'
]

scaler = MinMaxScaler()
df[features] = scaler.fit_transform(df[features])
```
