```python
import pandas as pd
import numpy as np

from google.colab import files

from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
df = pd.read_excel("Mldata.xlsx")
```

```python
print("Dataset:")
display(df)
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe(include="all")
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include="object").columns

print("Numerical columns:")
print(numeric_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
for column in categorical_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
print(df["GENDER"].unique())
```

```python
df["GENDER"] = df["GENDER"].astype("string").str.strip().str.lower()

df["GENDER"] = df["GENDER"].replace({
    "m": "Male",
    "male": "Male",
    "mens":"Male",
    "f": "Female",
    "female": "Female"
})

print(df["GENDER"].unique())
```

```python
print(df["GENDER"].value_counts(dropna=False))
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Duplicate rows after removal:", df.duplicated().sum())
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=["GENDER"],
    drop_first=True
)
```

```python
display(df_encoded.head())
```

```python
numeric_columns = df_encoded.select_dtypes(
    include=np.number
).columns

print(numeric_columns)
```

```python
for column in numeric_columns:

    Q1 = df_encoded[column].quantile(0.25)
    Q3 = df_encoded[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_encoded[
        (df_encoded[column] < lower_bound) |
        (df_encoded[column] > upper_bound)
    ]

    print(column, ":", len(outliers), "potential outliers")
```

```python
df_clean = df_encoded.copy()

for column in numeric_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[column] >= lower_bound) &
        (df_clean[column] <= upper_bound)
    ]
```

```python
print("Rows before outlier handling:", len(df_encoded))
print("Rows after outlier handling:", len(df_clean))
```

```python
standardized_data = df_clean.copy()

numeric_columns = standardized_data.select_dtypes(
    include=np.number
).columns

scaler = StandardScaler()

standardized_data[numeric_columns] = scaler.fit_transform(
    standardized_data[numeric_columns]
)

display(standardized_data.head())
```

```python
normalized_data = df_clean.copy()

numeric_columns = normalized_data.select_dtypes(
    include=np.number
).columns

minmax_scaler = MinMaxScaler()

normalized_data[numeric_columns] = minmax_scaler.fit_transform(
    normalized_data[numeric_columns]
)

display(normalized_data.head())
```

```python
print("Final Preprocessed Dataset:")
display(standardized_data)
```

```python
print("Final dataset shape:", standardized_data.shape)
```

```python
standardized_data.to_excel(
    "preprocessed_dataset.xlsx",
    index=False
)
```

```python
standardized_data.to_csv(
    "preprocessed_dataset.csv",
    index=False
)
```
